```python
import requests
import pandas as pd

# 1. Cities and their coordinates
cities = {
    "Marrakech": (31.63, -8.00),
    "Casablanca": (33.57, -7.59),
    "Rabat": (34.02, -6.84),
    "Tanger": (35.76, -5.83),
    "Agadir": (30.43, -9.60)
}

# 2. Empty list to store all cities' data
all_data = []

# 3. Get weather data for each city
for city, coordinates in cities.items():

    latitude = coordinates[0]
    longitude = coordinates[1]

    # API URL
    url = "https://api.open-meteo.com/v1/forecast"

    # Parameters
    params = {
        "latitude": latitude,
        "longitude": longitude,
        "hourly": "temperature_2m,relative_humidity_2m,wind_speed_10m",
        "forecast_days": 1
    }

    # Send request to API
    response = requests.get(url, params=params)

    data = response.json()

    hourly = data["hourly"]


    # Convert data to DataFrame
    df = pd.DataFrame(hourly)

    # Add city name
    df["city"] = city

    # Add DataFrame to list
    all_data.append(df)

# 4. Combine all cities into one DataFrame
weather_df = pd.concat(all_data, ignore_index=True)

# 5. Rename columns
weather_df = weather_df.rename(columns={
    "time": "datetime",
    "temperature_2m": "temperature_c",
    "relative_humidity_2m": "humidity_percent",
    "wind_speed_10m": "wind_speed_kmh"
})

# 6. Display the first rows
weather_df.shape

weather_df.to_csv("weather_data.csv", index=False)
```
